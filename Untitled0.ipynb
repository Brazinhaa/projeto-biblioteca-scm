```python
# 1. Função para cadastrar as notas e armazenar em uma lista
def cadastrar_notas():
    notas = [] # A atividade exige que as notas sejam armazenadas em uma lista
    print("Digite as notas do aluno. (Digite -1 para finalizar a inserção)")

    # Estrutura de repetição para permitir inserir múltiplas notas
    while True:
        try:
            nota = float(input("Insira uma nota (0 a 10): "))
            if nota == -1:
                break # Condição de parada do laço
            elif 0 <= nota <= 10:
                notas.append(nota) # Adiciona a nota na lista
            else:
                print("Nota inválida! Insira um valor entre 0 e 10.")
        except ValueError:
            print("Por favor, digite um número válido.")

    return notas

# 2. Função para calcular a média das notas
def calcular_media(notas):
    if len(notas) == 0:
        return 0
    # Calcula a soma das notas dividida pela quantidade de notas inseridas
    media = sum(notas) / len(notas)
    return media

# 3. Função para determinar a situação usando estrutura condicional
def determinar_situacao(media):
    # Se a média for maior ou igual a 7, aprovado. Menor que 7, reprovado.
    if media >= 7.0:
        return "Aprovado"
    else:
        return "Reprovado"

# 4. Função para exibir o relatório final
def relatorio_final():
    print("\n--- INÍCIO DO SISTEMA ---")
    lista_de_notas = cadastrar_notas()

    if len(lista_de_notas) > 0:
        media_aluno = calcular_media(lista_de_notas)
        situacao_aluno = determinar_situacao(media_aluno)

        # Exibir relatório conforme exigido
        print("\n=== RELATÓRIO FINAL ===")
        print(f"Notas inseridas: {lista_de_notas}")
        print(f"Média calculada: {media_aluno:.2f}")
        print(f"Situação do aluno: {situacao_aluno}")
    else:
        print("\nNenhuma nota foi inserida. Sistema encerrado.")

# Executar o sistema principal
relatorio_final()
```
